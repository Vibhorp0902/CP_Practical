```python
C = 2.5
k = 0.04
t = 12
dt = 1  # time step

dcdt = - k * C
C_new = C + dcdt * dt
print(f"C = {C_new}")
conversion = (2.5 - C) * 100


for i in range(t+1):
    dcdt = - k * C
    print(f"At time {i} minutes, C = {C:.4f}, Conversion: {conversion:.2f}%")
    C = C + dcdt * dt
    conversion = (2.5 - C) * 100
print(f"final concentration after 10 minutes: {C:.4f}")
```

A laboratary has n candidate solvents and want to test r of them together in a mixed solvent . Write a recursive function factorial (n) and a function combination (n,r) n!/(r!(n-r)!) that uses it.Reject invalid input . Print the no. of posssible mixtures for n=9 and r = 3 and show the message for n = 6 , r = 8


```python
def factorial(n):
    if not isinstance(n, int) or isinstance(n, bool) or n < 0:
        raise ValueError("n must be a non-negative integer")

    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)


def combination(n, r):
    if (not isinstance(n, int) or isinstance(n, bool) or
            not isinstance(r, int) or isinstance(r, bool)):
        raise ValueError("n and r must be integers")
    if n < 0 or r < 0:
        raise ValueError("n and r must be non-negative")
    if r > n:
        raise ValueError("r cannot be greater than n")

    return factorial(n) // (factorial(r) * factorial(n - r))
# 9 candidate solvents, selecting 3 together
print("Possible mixtures:", combination(9, 3))

# Invalid case
try:
    print(combination(6, 8))
except ValueError as error:
    print("Error:", error)
```

During a crystallisation run samples are tken at times = fibonacci number * 4 minutes . Write an iterative function fib_iter(n) that returns a list of the first n fibonacci number and a recursive function fib_rec(k) that returns the kth fibonacci number . For n = 10 check that both functions agree and print a table : sample no. |Fibonacci tag | Sampling time.

```python
def fib_iter(n):
    if not isinstance(n, int) or isinstance(n, bool) or n < 0:
        raise ValueError("n must be a non-negative integer")

    sequence = []
    a, b = 0, 1

    for _ in range(n):
        sequence.append(a)
        a, b = b, a + b

    return sequence


def fib_rec(k):
    if not isinstance(k, int) or isinstance(k, bool) or k < 0:
        raise ValueError("k must be a non-negative integer")
    if k == 0:
        return 0
    if k == 1:
        return 1
    return fib_rec(k - 1) + fib_rec(k - 2)


n = 10
iterative_values = fib_iter(n)
recursive_values = [fib_rec(k) for k in range(n)]

# Check that both implementations agree
if iterative_values != recursive_values:
    raise RuntimeError("Iterative and recursive Fibonacci values do not agree")

print("Sample no. | Fibonacci tag | Sampling time")
print("---------- | ------------ | -------------")
for k, fibonacci_tag in enumerate(iterative_values):
    sampling_time = fibonacci_tag * 4
    print(f"{k + 1:<10} | {fibonacci_tag:<12} | {sampling_time} minutes")
```

I have a chemical batch identifier such as DIST2026X9Q3. Explain how a Python for loop can inspect each character and identify which characters are digits. Mention the role of isdigit(). Do not write the complete program.

```python
batch_id = "DIST2026X9Q3"

for character in batch_id:
    if character.isdigit():
        print(character)

print("Number of digits in batch_id: ", sum(c.isdigit() for c in batch_id))
sum_of_digits = 0
for char in batch_id:
    if char.isdigit():
        sum_of_digits += int(char)
print("Sum of digits in batch number:", sum_of_digits)
```

Ethylene gas is slowly compressed in piston cylinder at constant temperature 
volume 1 ,2 ,3 ,4 ,5  pressure 5.03 , 2.52 ,1.65 ,1.26 ,1.01 .Plot pressure vs volume as shatter plot  and calculate k as mean of P x V and draw the reference curve P = k / v on same graph

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

volume = ([1, 2, 3, 4, 5])
pressure = np.array([5.03, 2.52, 1.65, 1.26, 1.01])

# Create DataFrame
data = pd.DataFrame({
    "Volume": volume,
    "Pressure": pressure
})

# Mean of P × V
data["P_times_V"] = data["Pressure"] * data["Volume"]
k = data["P_times_V"].mean()

print(data)
print(f"\nk (mean of P × V) = {k:.4f}")
# Smooth reference curve
v_curve = np.linspace(volume[0], volume[-1], 300)
p_curve = k / v_curve

# Scatter plot and reference curve
plt.scatter(volume, pressure, color="blue", label="Measured data")
plt.plot(v_curve, p_curve, color="red", linestyle="--",
         label=f"Reference: P = {k:.4f}/V")
plt.xlabel("Volume")
plt.ylabel("Pressure")
plt.title("Ethylene Gas Compression")
plt.legend()
plt.grid(True)
plt.show()
```
