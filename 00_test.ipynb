```python
# Create a simple if elif else code to check marks and give grade A B C D E based on the marks entered by the user.
marks = float(input("Enter the marks: "))

if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
else:
    grade = "E"

print(f"The grade is: {grade}")
```
